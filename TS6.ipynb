# Introducción teórica

La respuesta en frecuencia de un circuito permite analizar cómo varían el módulo y la fase de su función de transferencia en función de la frecuencia de la señal de entrada. Según su distribucion y los componentes, un circuito puede comportarse como un filtro pasa bajos, pasa altos, pasa banda o elimina banda.

Se estudian tres configuraciones de circuitos RLC, constituidos por resistencias, inductores y capacitores. 
Las configuraciones analizadas son: 
A) Un circuito RLC en serie con la salida medida sobre el capacitor
B) Una resistencia en paralelo con un capacitor, conectada en serie con un inductor, con la salida medida sobre el inductor
c) Una resistencia en serie con un conjunto formado por un inductor y un capacitor en paralelo, con la salida medida sobre dicho conjunto.

Para cada circuito se obtiene la función de transferencia H(s), definida como el cociente entre la transformada de Laplace de la tensión de salida y la transformada de Laplace de la tensión de entrada, suponiendo condiciones iniciales nulas. Para ello, se utilizan las impedancias de los componentes en el dominio de Laplace y el principio del divisor de tensión, considerando las asociaciones serie y paralelo correspondientes.

Una vez obtenida cada función de transferencia, se determina la respuesta en frecuencia mediante la sustitución  s=jw  donde  w es la frecuencia angular. A partir de esta expresión, se calculan el módulo y la fase de cada sistema. 



### Cálculo del ancho de banda

Se calculan las frecuencias de corte, definidas como aquellas en las que el módulo de la función de transferencia disminuye a $1/\sqrt{2}$, lo que corresponde a una caída aproximada de $3$ dB.

Para los filtros pasa bajos y pasa altos, se determina una frecuencia de corte $\omega_c$ mediante la condición:

$$
|H(j\omega_c)|=\frac{|H|_{\mathrm{ref}}}{\sqrt{2}}
$$


Para el filtro pasa banda, se determinan dos frecuencias de corte, $\omega_1$ y $\omega_2$. El ancho de banda se calcula como la diferencia entre ambas frecuencias:

$$
BW=\omega_2-\omega_1
$$

A partir de estas frecuencias, se caracteriza la banda de paso de cada circuito y se identifica su comportamiento como filtro.




# Resultados

Se analizaron las tres configuraciones utilizando los siguientes parámetros de diseño:

$$
\omega_0=1\ \text{rad/s},\qquad
Q=\frac{\sqrt{2}}{2},\qquad
R=1\ \Omega
$$

## Circuito A: RLC en serie con salida sobre el capacitor

Para obtener los parámetros del circuito se utilizaron las relaciones:

$$
\omega_0=\frac{1}{\sqrt{LC}},
\qquad
Q=\frac{1}{R}\sqrt{\frac{L}{C}}
$$

Con los valores establecidos, se obtiene:

$$
L=\frac{\sqrt{2}}{2}\ \text{H},
\qquad C=\sqrt{2}\ \text{F}
$$

La función de transferencia resulta:

$$
H_A(s)=\frac{1}{s^2+\sqrt{2}s+1}
$$

Reemplazando $s=j\omega$, se obtiene la respuesta en frecuencia:

$$
H_A(j\omega)=
\frac{1}{1-\omega^2+j\sqrt{2}\omega}
$$

El módulo es:

$$
|H_A(j\omega)|=
\frac{1}{\sqrt{(1-\omega^2)^2+2\omega^2}}
$$

La fase es:

$$
\varphi_A(\omega)=
-\operatorname{atan2}
\left(\sqrt{2}\omega,1-\omega^2\right)
$$

Este circuito presenta un comportamiento de filtro pasa bajos.

## Circuito B: Resistencia y capacitor en paralelo, en serie con el inductor

Para esta configuración se utilizan las relaciones:

$$
\omega_0=\frac{1}{\sqrt{LC}},
\qquad
Q=R\sqrt{\frac{C}{L}}
$$

Los valores obtenidos son:

$$
L=\sqrt{2}\ \text{H},
\qquad
C=\frac{\sqrt{2}}{2}\ \text{F}
$$

La función de transferencia es:

$$
H_B(s)=
\frac{s^2+\sqrt{2}s}
{s^2+\sqrt{2}s+1}
$$

La respuesta en frecuencia queda expresada como:

$$
H_B(j\omega)=
\frac{-\omega^2+j\sqrt{2}\omega}
{1-\omega^2+j\sqrt{2}\omega}
$$

El módulo es:

$$
|H_B(j\omega)|=
\frac{\sqrt{\omega^4+2\omega^2}}
{\sqrt{(1-\omega^2)^2+2\omega^2}}
$$

La fase es:

$$
\begin{aligned}
\varphi_B(\omega)={}&
\operatorname{atan2}(\sqrt{2}\omega,-\omega^2)\\
&-\operatorname{atan2}(\sqrt{2}\omega,1-\omega^2)
\end{aligned}
$$

Este circuito presenta un comportamiento de filtro pasa altos.

## Circuito C: Resistencia en serie con un conjunto LC en paralelo

Para esta configuración también se utilizan las relaciones:

$$
\omega_0=\frac{1}{\sqrt{LC}},
\qquad
Q=R\sqrt{\frac{C}{L}}
$$

Los valores obtenidos son:

$$
L=\sqrt{2}\ \text{H},
\qquad
C=\frac{\sqrt{2}}{2}\ \text{F}
$$

La función de transferencia es:

$$
H_C(s)=
\frac{\sqrt{2}s}
{s^2+\sqrt{2}s+1}
$$

La respuesta en frecuencia resulta:

$$
H_C(j\omega)=
\frac{j\sqrt{2}\omega}
{1-\omega^2+j\sqrt{2}\omega}
$$

El módulo es:

$$
|H_C(j\omega)|=
\frac{\sqrt{2}\omega}
{\sqrt{(1-\omega^2)^2+2\omega^2}}
$$

La fase es:

$$
\varphi_C(\omega)=
\frac{\pi}{2}-
\operatorname{atan2}
\left(\sqrt{2}\omega,1-\omega^2\right)
$$

Este circuito presenta un comportamiento de filtro pasa banda.

## Comparación de los resultados

Las tres configuraciones presentan diferentes respuestas en frecuencia:

- **Circuito A:** pasa bajos; atenúa las frecuencias altas.
- **Circuito B:** pasa altos; atenúa las frecuencias bajas.
- **Circuito C:** pasa banda; atenúa las frecuencias bajas y altas, con un máximo en torno a la frecuencia de resonancia.

En los tres casos, la frecuencia natural es $\omega_0=1\ \text{rad/s}$ y el factor de calidad es $Q=\sqrt{2}/2$. Los diagramas de Bode permiten comparar el módulo y la fase de cada configuración.


# Tipo de filtro y cálculo del ancho de banda

## Circuito A: RLC en serie con salida sobre el capacitor

La función de transferencia es:

$$
H_A(s)=\frac{1}{s^2+\sqrt{2}s+1}
$$

El módulo de la respuesta en frecuencia es:

$$
|H_A(j\omega)|=\frac{1}{\sqrt{1+\omega^4}}
$$

Para determinar el tipo de filtro, se analizan los límites del módulo:

$$
\lim_{\omega\to 0}|H_A(j\omega)|=1
$$

$$
\lim_{\omega\to\infty}|H_A(j\omega)|=0
$$

El circuito se comporta como un **filtro pasa bajos**, debido a que permite el paso de las bajas frecuencias y atenúa las altas.

 La frecuencia de corte $\omega_c$ 

$$
|H_A(j\omega_c)|=\frac{1}{\sqrt{2}}
$$



Por lo tanto:

$$
\boxed{\omega_c=1\ \text{rad/s}}
$$

Su ancho de banda  es:

$$
\boxed{BW=\omega_c=1\ \text{rad/s}}
$$

## Circuito B: Resistencia y capacitor en paralelo, en serie con el inductor

La función de transferencia es:

$$
H_B(s)=\frac{s^2+\sqrt{2}s}{s^2+\sqrt{2}s+1}
$$

El módulo de la respuesta en frecuencia es:

$$
|H_B(j\omega)|=
\sqrt{\frac{\omega^4+2\omega^2}{1+\omega^4}}
$$

Analizando los límites:

$$
\lim_{\omega\to 0}|H_B(j\omega)|=0
$$

$$
\lim_{\omega\to\infty}|H_B(j\omega)|=1
$$

Por lo tanto, el circuito se comporta como un **filtro pasa altos**, ya que atenúa las bajas frecuencias y permite el paso de las altas.

La frecuencia de corte $\omega_c$ se calcula mediante:

$$
|H_B(j\omega_c)|=\frac{1}{\sqrt{2}}
$$

Reemplazando:

$$
\sqrt{\frac{\omega_c^4+2\omega_c^2}
{1+\omega_c^4}}=\frac{1}{\sqrt{2}}
$$

Elevando al cuadrado:

$$
\frac{\omega_c^4+2\omega_c^2}
{1+\omega_c^4}=\frac{1}{2}
$$

Multiplicando y reordenando:

$$
2\omega_c^4+4\omega_c^2=1+\omega_c^4
$$

$$
\omega_c^4+4\omega_c^2-1=0
$$



$$
\boxed{\omega_c=\sqrt{\sqrt{5}-2}
\approx0.486\ \text{rad/s}}
$$

$\omega_c$ sin un límite superior, su ancho de banda ideal es infinito:

$$
\boxed{BW\to\infty}
$$



## Circuito C: Resistencia en serie con un conjunto LC en paralelo

La función de transferencia es:

$$
H_C(s)=\frac{\sqrt{2}s}{s^2+\sqrt{2}s+1}
$$

El módulo de la respuesta en frecuencia es:

$$
|H_C(j\omega)|=\frac{\sqrt{2}\omega}{\sqrt{1+\omega^4}}
$$

Analizando los límites:

$$
\lim_{\omega\to 0}|H_C(j\omega)|=0
$$

$$
\lim_{\omega\to\infty}|H_C(j\omega)|=0
$$

El módulo alcanza su máximo en $\omega_0=1\ \text{rad/s}$

$$
|H_C(j1)|=1
$$

Por lo tanto, el circuito se comporta como un **filtro pasa banda**, ya que atenúa las bajas y las altas frecuencias y permite el paso de una banda intermedia.

Las frecuencias de corte $\omega_1$ y $\omega_2$

$$
|H_C(j\omega)|=\frac{1}{\sqrt{2}}
$$

Reemplazando:

$$
\frac{\sqrt{2}\omega}{\sqrt{1+\omega^4}}
=\frac{1}{\sqrt{2}}
$$

Elevando al cuadrado:

$$
\frac{2\omega^2}{1+\omega^4}=\frac{1}{2}
$$

Reordenando:

$$
\omega^4-4\omega^2+1=0
$$

Se obtienen las dos soluciones 

$$
\boxed{\omega_1=\sqrt{2-\sqrt{3}}
\approx0.518\ \text{rad/s}}
$$

$$
\boxed{\omega_2=\sqrt{2+\sqrt{3}}
\approx1.932\ \text{rad/s}}
$$

El ancho de banda se calcula como la diferencia entre ambas frecuencias de corte

$$
BW=\omega_2-\omega_1
$$

Reemplazando

$$
BW\approx1.932-0.518
$$


$$
\boxed{BW\approx1.414\ \text{rad/s}}
$$



## Conclusión


- **Circuito A:** filtro pasa bajos, con $BW=1\ \text{rad/s}$.
- **Circuito B:** filtro pasa altos, con $\omega_c\approx0.486\ \text{rad/s}$ y ancho de banda ideal infinito.
- **Circuito C:** filtro pasa banda, con $\omega_1\approx0.518\ \text{rad/s}$, $\omega_2\approx1.932\ \text{rad/s}$ y $BW\approx1.414\ \text{rad/s}$.
