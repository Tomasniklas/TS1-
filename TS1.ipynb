# Ejercicio 1


Se genera:

1. Una señal sinusoidal de $2\,kHz$ que tenga al menos 10 puntos por período.
2. La misma señal con una potencia media de $2\,W$ y un desfase de $\pi/2$.
3. Una secuencia aleatoria de ruido normalmente distribuido con valor medio $0\,V$ y varianza $0.1$.
4. Una secuencia aleatoria de ruido uniformemente distribuido con valor medio $0\,V$ y varianza $0.1$.
5. Un pulso rectangular de $2\,kHz$, potencia media $1\,W$ y ciclo de actividad del $50\%$.



# Introducción 

Para la generación y análisis de señales discretas se utilizan diferentes parámetros, entre ellos la amplitud, el valor medio, la varianza y la potencia media.

Para una señal discreta $x[n]$ de $N$ muestras, la potencia media se calcula como:

$$
P = \frac{1}{N}\sum_{n=0}^{N-1}|x[n]|^2
$$



## Señales sinusoidales

Una señal sinusoidal puede expresarse como:

$$
x[n] = A\sin(2\pi f_0 nT_s+\phi)
$$

donde $A$ es la amplitud, $f_0$ la frecuencia, $T_s$ el período de muestreo y $\phi$ la fase.

Para una señal sinusoidal, la potencia media está relacionada con su amplitud de forma:

$$
P = \frac{A^2}{2}
$$

Por lo tanto, si se desea obtener una potencia, la amplitud puede calcularse como:

$$
A = \sqrt{2P}
$$


Para una potencia de $2\,W$ entonces queda:

$$
A = \sqrt{2(2)} = 2\,V
$$

## Relación entre potencia y varianza

La potencia media de una señal puede relacionarse con su valor medio y su varianza mediante:

$$
P = \sigma^2 + \mu^2
$$

donde $\mu$ es el valor medio y $\sigma^2$ es la varianza.

Por lo tanto, cuando la señal tiene valor medio cero:

$$
\mu = 0
$$

la potencia queda directamente relacionada con la varianza:

$$
P = \sigma^2
$$


## Señal rectangular

Para una señal rectangular bipolar con amplitudes $+A$ y $-A$, la potencia media es:

$$
P = A^2
$$

Por lo tanto, para obtener una potencia de $1\,W$:

$$
A = \sqrt{1} = 1\,V
$$

Además, un ciclo de actividad del $50\%$ significa que la señal permanece durante la mitad del período en el nivel positivo y durante la otra mitad en el nivel negativo.

## Frecuencia de muestreo

La señal a generar tiene una frecuencia de:

$$
f_0 = 2\,kHz
$$

Su período es:

$$
T_0 = \frac{1}{f_0} = \frac{1}{2000} = 0.5\,ms
$$

Como se requieren al menos 10 muestras por período, se selecciona:

$$
f_s = 10f_0
$$

por lo tanto:

$$
f_s = 10(2000) = 20000\,Hz = 20\,kHz
$$

```python
# -*- coding: utf-8 -*-
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# CONSTANTES
N = 1000  # Cantidad de muestras
ff = 2000  # Frecuencia de 2 kHz


fs = 20000  # Frecuencia de muestreo (10 puntos)

# VECTOR TEMPORAL
tt = np.arange(N) / fs

# 1. SEÑAL SINUSOIDAL DE 2 kHz
xx1 = np.sin(2 * np.pi * ff * tt)

# 2. SEÑAL SINUSOIDAL DE 2W Y FASE π/2

xx2 = 2 * np.sin(2 * np.pi * ff * tt + np.pi/2)

# 3. RUIDO NORMAL (media 0, varianza 0.1)
xx3 = np.sqrt(0.1) * np.random.randn(N)

# 4. RUIDO UNIFORME (media 0, varianza 0.1)

a = np.sqrt(3 * 0.1)
xx4 = np.random.uniform(-a, a, N)

# 5. PULSO RECTANGULAR (2 kHz, 1W, duty 50%)

xx5 = signal.square(2 * np.pi * ff * tt, duty=0.5)

# GRAFICAR
plt.figure(figsize=(12, 10))

plt.subplot(5, 1, 1)
plt.plot(tt, xx1)
plt.title('1. Señal sinusoidal 2 kHz')
plt.ylabel('Amplitud [V]')
plt.grid()

plt.subplot(5, 1, 2)
plt.plot(tt, xx2)
plt.title('2. Señal sinusoidal 2 kHz, 2W, fase π/2')
plt.ylabel('Amplitud [V]')
plt.grid()

plt.subplot(5, 1, 3)
plt.plot(tt, xx3)
plt.title('3. Ruido normal (media 0, varianza 0.1)')
plt.ylabel('Amplitud [V]')
plt.grid()

plt.subplot(5, 1, 4)
plt.plot(tt, xx4)
plt.title('4. Ruido uniforme (media 0, varianza 0.1)')
plt.ylabel('Amplitud [V]')
plt.grid()

plt.subplot(5, 1, 5)
plt.plot(tt, xx5)
plt.title('5. Pulso rectangular 2 kHz, 1W, duty 50%')
plt.xlabel('Tiempo [s]')
plt.ylabel('Amplitud [V]')
plt.grid()

plt.tight_layout()
plt.show()
```
